```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np


beta = 10
omega = -2

# Generate a range of x-values 
x = np.linspace(-10, 10, 100)

# Calculate y-values and assign to y1
y1 = omega * x + beta

# Plot the 1D linear function
plt.figure(figsize=(8, 5))
plt.plot(x, y1, label=f"y = {omega}x + {beta}", color="orange")
plt.title("1D Linear Function")
plt.xlabel("x")
plt.ylabel("y")
plt.axhline(0, color="black", linewidth=0.5, linestyle="--")
plt.axvline(0, color="black", linewidth=0.5, linestyle="--")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Define parameters
beta = -3
omega_1 = 1
omega_2 = 0

# Generate a grid of X1 and X2 values
x1 = np.linspace(-10, 10, 100)
x2 = np.linspace(-10, 10, 100)
X1, X2 = np.meshgrid(x1, x2)

# Calculate the 3D linear surface and assign to Y1
Y1 = omega_1 * X1 + omega_2 * X2 + beta

# Plot the 3D surface
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection="3d")

surface = ax.plot_surface(X1, X2, Y1, cmap="viridis", edgecolor="none", alpha=0.8)

# Set labels and title
ax.set_title("3D Linear Surface (Y1 = X1 - 3)")
ax.set_xlabel("X1")
ax.set_ylabel("X2")
ax.set_zlabel("Y1")

# Add a colorbar for reference
fig.colorbar(surface, ax=ax, shrink=0.5, aspect=5)

plt.show()
```
