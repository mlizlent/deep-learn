```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

inputs = pd.DataFrame({
    "Water (component 4)(kg in a m^3 mixture)": np.random.normal(180, 20, 500)
})

water_feature = inputs["Water (component 4)(kg in a m^3 mixture)"]

plt.hist(water_feature, bins=20)
plt.xlabel("Water (kg in a m^3 mixture)")
plt.ylabel("Frequency")
plt.title("Distribution of Water Content")
plt.show()
```
